```python
# https://leetcode.com/problems/trapping-rain-water/description/from typing import Optional
from typing import List

def make_case(maxh, case, height):
    test = []
    for i in range(len(height)):
        temp = height[i]-case
        if temp > 0:
            test.append(1)
        else:
            test.append(0)
    return test

def count(test):
    frontzeros = 0
    endzeros = 0
    for i in range(len(test)):
        if test[i] == 0:
            frontzeros+=1
        else:
            break
    for i in range(len(test)):
        if test[len(test)-1-i] == 0:
            endzeros+=1
        else:
            break
    return (len(test)-sum(test))-frontzeros-endzeros

def trap(height: List[int]) -> int:
    maxh = max(height)
    cnt = 0

    visited = {}
    for i in range(maxh):
        test = tuple(make_case(maxh, i, height))
        if test in visited:
            cnt+=visited[test]
        else:
            a = count(test)
            visited[test]=a
            cnt+=a
        # print(test, cnt)
        
    return cnt
```
