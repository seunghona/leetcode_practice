### EASY

```python
# https://leetcode.com/problems/climbing-stairs/
def climbStairs(n: int) -> int:
    memo = [0] * (n + 1)

    def count(i, memo):
        if i == 1:
            memo[i]=1
            return 1, memo
        if i == 2:
            memo[i]=2
            return 2, memo
        if memo[i] != 0:
            return memo[i], memo
        
        memo[i] = count(i - 1, memo)[0] + count(i - 2, memo)[0]
        return memo[i], memo
    ans, memo = count(n, memo)
    # print(memo)
    return ans
```

```python
# https://leetcode.com/problems/min-cost-climbing-stairs/
from typing import List
def minCostClimbingStairs(cost: List[int]) -> int:
    n = len(cost)
    acc = [0]*(n)
    for i in range(2, n):
        acc[i] = min(acc[i-1]+cost[i-1], acc[i-2]+cost[i-2])
    return min(acc[-1]+cost[-1], acc[-2]+cost[-2])
```

### INTERMEDIATE

```python
# https://leetcode.com/problems/house-robber/
from typing import List

def rob(nums: List[int]) -> int:
    n=len(nums)
    if n==1:
        return nums[0]
    acc=[0]*n
    acc[0]=nums[0]
    acc[1]=max(nums[1], nums[0])
    for i in range(2, n):
        acc[i]=max(acc[i-1], acc[i-2]+nums[i])
    return acc[-1]
```

```python
nums = [1,2,3]
rob(nums)
```

```python
# https://leetcode.com/problems/house-robber-ii/
def rob(nums: List[int]) -> int:
    def rob_(nums):
        n=len(nums)
        if n==1:
            return nums[0]
        acc=[0]*n
        acc[0]=nums[0]
        acc[1]=max(nums[1], nums[0])
        for i in range(2, n):
            acc[i]=max(acc[i-1], acc[i-2]+nums[i])
        return acc[-1]
    n=len(nums)
    if n<3:
        return max(nums)
    nums1 = nums[0:-1]
    nums2 = nums[1:]
    # print(nums1, nums2)
    # print(rob_(nums1))
    # print(rob_(nums2))
    return max(rob_(nums1), rob_(nums2))
```

```python
nums = [1,2,3,1]
rob(nums)
```

```python
# https://leetcode.com/problems/coin-change/description/
from typing import List

def coinChange(coins: List[int], amount: int) -> int:
    n=amount
    if n==0:
        return 0
    if n<min(coins):
        return -1
    combos = [-1]*(n+1)
    coins.sort()
    for i in coins:  
        if i<=n:
            combos[i]=1
    for i in range(coins[0], len(combos)):
        for coin in coins:
            if combos[i]==-1 and i-coin>0:
                if combos[i-coin]>0:
                    combos[i]=combos[i-coin]+1
            elif combos[i]>0 and i-coin>0 and combos[i-coin]>0:
                if combos[i]>combos[i-coin]+1:
                    combos[i]=combos[i-coin]+1
    # print(combos)
    return combos[n]
    
```

```python
coins = [186,419,83,408]
coinChange(coins, 6249)
```

```python
# https://leetcode.com/problems/unique-paths/
def uniquePaths(m: int, n: int) -> int:
    dp = []
    for i in range(m):
        dp.append([0]*n)
    dp[0][0]=1
    # print(dp)
    for i in range(m):
        dp[i][0]=1
    for i in range(n):
        dp[0][i]=1
    # print(dp)
    for i in range(1, m):
        for j in range(1, n):
            dp[i][j]=dp[i-1][j]+dp[i][j-1]
    # print(dp)
    return dp[m-1][n-1]
```
