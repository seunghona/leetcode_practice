```python
# https://leetcode.com/problems/determine-if-two-strings-are-close/description/?envType=study-plan-v2&envId=leetcode-75
def closeStrings(word1: str, word2: str) -> bool:
    if len(word1)!=len(word2):
        return False
    # s1 = set(word1)
    # s2 = set(word2)
    # if s1!=s2:
    #     return False
    # s1cnt = {}
    # s2cnt = {}
    # for i in range(len(word1)):
    #     if word1[i] in s1cnt.keys():
    #         s1cnt[word1[i]]+=1
    #     else:
    #         s1cnt[word1[i]]=1
    # for i in range(len(word2)):
    #     if word2[i] in s2cnt.keys():
    #         s2cnt[word2[i]]+=1
    #     else:
    #         s2cnt[word2[i]]=1
    
    s1cnt=Counter(word1)
    s2cnt=Counter(word2)
    if set(s1cnt.keys())!=set(s2cnt.keys()):
        return False
    a = list(s1cnt.values())
    a.sort()
    b = list(s2cnt.values())
    b.sort()
    for i in range(len(a)):
        if a[i]!=b[i]:
            return False
    return True
```

```python
# word1 = "abbzzca"
# word2 = "babzzcz"
word1 = 'abc'
word2 = 'bca'
print(closeStrings(word1, word2))
```
