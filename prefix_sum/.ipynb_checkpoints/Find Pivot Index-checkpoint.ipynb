```python
# https://leetcode.com/problems/find-pivot-index/description/?envType=study-plan-v2&envId=leetcode-75
from typing import List
def pivotIndex(nums: List[int]) -> int:
    if len(nums)==1:
        return 0
    total = sum(nums)
    cur = 0
    for i in range(nums):
        cur+=nums[i]
        if cur*2==total:
            return i+1
```
