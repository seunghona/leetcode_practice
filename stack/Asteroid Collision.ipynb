```python
# https://leetcode.com/problems/asteroid-collision/description/?envType=study-plan-v2&envId=leetcode-75
from typing import List
def asteroidCollision(asteroids: List[int]) -> List[int]:
    ans=[]
    for i in range(len(asteroids)):
        now=asteroids[i]
        if len(ans)==0 or now>0:
            ans.append(now)
        else:
            if ans[-1]<0:
                ans.append(now)
            else:
                print(ans, now)
                while True:
                    if ans[-1]<0:
                        ans.append(now)
                        break
                    if abs(now)<ans[-1]:
                        break
                    elif abs(now)==ans[-1]:
                        ans.pop(-1)
                        break
                    else:
                        ans.pop(-1)
                        if len(ans)==0:
                            ans.append(now)
                            break
    return ans
```

```python
# asteroids = [-2,-1,1,2]
# asteroids = [10,2,-5]
# asteroids = [1,-2,-2,-2]
asteroids = [-2,-2,1,-2]
print(asteroidCollision(asteroids))
```
