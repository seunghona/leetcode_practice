```python
# https://leetcode.com/problems/decode-string/description/?envType=study-plan-v2&envId=leetcode-75
def decodeString(s: str) -> str:
    stack = []
    pre=''
    cnt=''
    for i in range(len(s)):
        now = s[i]
        if now.isdigit():
            cnt+=now
        elif now=='[':
            stack.append(pre)
            stack.append(int(cnt))
            pre=''
            cnt=''
        elif now==']':
            word=pre
            cnt=stack.pop()
            pre=stack.pop()
            pre=pre+cnt*word
            cnt=''
        else:
            pre+=now
        print(stack, pre)
    return pre
```

```python
# s="3[a]2[bc]"
# s='3[a2[c]]'
s='100[leetcode]'
print(decodeString(s))
```
