```python
# https://leetcode.com/problems/removing-stars-from-a-string/?envType=study-plan-v2&envId=leetcode-75
def removeStars(s: str) -> str:
    idx=0
    while True:
        if s[idx]=='*':
            idx-=1
            s.pop(idx)
            s.pop(idx)
        else:
            idx+=1
        if idx==len(s):
            break
    return s
```
