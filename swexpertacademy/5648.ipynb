```python
import sys

direcs = {0:(0,0.5), 1:(0,-0.5), 2:(-0.5,0), 3:(0.5,0)}

class Atom:
    def __init__(self, pos, num, direc, state):
        self.pos = pos
        self.num = num
        self.direc = direc
        self.state = state
    def move(self):
        self.pos=(self.pos[0]+direcs[self.direc][0], self.pos[1]+direcs[self.direc][1])
        
def solve(N, atoms):
    energy=0
    deadcnt = 0
    for i in range(4000):
        if deadcnt==len(atoms):
            break
        pos_sum={}
        
        for atom in atoms:
            if atom.state==0:
                continue
            atom.move()
            if atom.pos not in pos_sum:
                pos_sum[atom.pos]=[atom]
            else:
                pos_sum[atom.pos].append(atom)

        for pos in pos_sum.keys():
            if len(pos_sum[pos])>1:
                this_energy=0
                for j in range(len(pos_sum[pos])):
                    this_energy+=pos_sum[pos][j].num
                    pos_sum[pos][j].state=0
                    deadcnt+=1
                energy+=this_energy
        
    return energy
        

sys.stdin = open("sample_input_5648.txt", "r")
T = int(sys.stdin.readline())
# T = int(input())
# 여러개의 테스트 케이스가 주어지므로, 각각을 처리합니다.
# print(T)
for test_case in range(1, T + 1):
    N = int(sys.stdin.readline())
    atoms = []
    for i in range(N):
        temp = list(map(int, sys.stdin.readline().split()))
        # print(temp)
        pos = (temp[0], temp[1])
        direc = temp[2]
        k = temp[3]
        atoms.append(Atom(pos, k, direc, 1))
    ans = solve(N, atoms)
    print('#{} {}'.format(test_case, ans))
    # break
```
