```python
import sys

class Person:
    def __init__(self, dist, state):
        self.dist = dist
        self.state = state
        # 1 moving, 2 wait, 3 stairs, 0 done
    def printp(self):
        print(self.dist, self.state)

def check(dists, stairval):
    if len(dists)==0:
        return 0
    dists.sort()
    # print(dists, stairval)
    persons = []
    t=dists[0]
    left=len(dists)
    for dist in dists:
        persons.append(Person(dist-t, 1))
    
    while left:
        instairs = 0
        for i in range(len(persons)):
            if persons[i].state==3:
                instairs+=1
        # now people in stairs known
        
        for i in range(len(persons)):
            person=persons[i]
            # person.printp()
        #     if person.dist==0 and person.state==1:
        #         if instairs<3:
        #             person.state=3
        #             person.dist=stairval
        #             instairs+=1
        #     elif person.state==1:
        #         person.dist-=1
        #     elif person.state==3:
        #         if person.dist>1:##### 0
        #             person.dist-=1
        #         else:
        #             person.state=0
        #             left-=1
        #     elif person.state==0:
        #         continue

            if person.dist==0 and person.state==1:
                person.state = 2
            elif person.state==1:
                person.dist-=1
            elif person.state==2:
                if instairs<3:
                    person.state=3
                    person.dist=stairval
                    instairs+=1
            elif person.state==3:
                if person.dist>1:##### 0
                    person.dist-=1
                else:
                    person.state=0
                    instairs-=1
                    left-=1
            elif person.state==0:
                continue
        t+=1
    # print('ans is {}'.format(t))
    return t-1
    
def solve(N, people, stairs):
    ans = float('inf')
    # for i in range(2**(len(people)-1)-3, 2**len(people)):
    for i in range(2**len(people)):
        destinations = str(format(i, '0'+str(len(people))+'b'))
        # print(destinations)
        s0 = []
        s1 = []
        for j in range(len(people)):
            if destinations[j]=='0':
                s0.append(abs(people[j][0]-stairs[0][0])+abs(people[j][1]-stairs[0][1]))
            else:
                s1.append(abs(people[j][0]-stairs[1][0])+abs(people[j][1]-stairs[1][1]))

        temp = max(check(s0, stairs[0][2]), check(s1, stairs[1][2]))
        ans = min(temp, ans)
        # break
        
    return ans
        
sys.stdin = open("sample_input_2383.txt", "r")
T = int(sys.stdin.readline())

for test_case in range(1, T + 1):
    N = int(sys.stdin.readline())
    people = []
    stairs = []
    for i in range(N):
        temp = list(map(int, sys.stdin.readline().split()))
        for j in range(len(temp)):
            val = temp[j]
            if val==0:
                continue
            elif val==1:
                people.append((i,j))
            else:
                stairs.append((i,j,val))
    # if test_case!=9:
    #     continue
    ans = solve(N, people, stairs)
    print('#{} {}'.format(test_case, ans))
    # break
```
