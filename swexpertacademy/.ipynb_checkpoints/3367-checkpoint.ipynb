```python
import sys

def get_seps(str1, size):
    ones=size
    front_zeros=0
    for i in range(len(str1)):
        if str1[i]=='0':
            front_zeros+=1
        else:
            break
    pairs=[]
    for i in range(1, ones):
        num1 = '0'*front_zeros+'1'*i+'0'*(len(str1)-i-front_zeros)
        num2 = '0'*(front_zeros+i)+'1'*(ones-i)+'0'*(len(str1)-ones-front_zeros)
        pairs.append((num1, num2))
    return pairs

def solve_old(N, parts):
    if N==1:
        return parts[0]
    elif N==2:
        return sum(parts)
        
    dp = {}
    history = {}
    # dp with only one part
    for i in range(len(parts)):
        base = '0'*N
        now = base[0:i]+'1'+base[i+1:]
        dp[now] = parts[i]
        history[now]=0

    for i in range(2, N+1):
        size=i
        for j in range(N-i+1):
            this = '0'*j + '1'*size + '0'*(N-j-size)
            pairs = get_seps(this, size)
            for pair in pairs:
                if this not in dp.keys():
                    dp[this]=dp[pair[0]]+dp[pair[1]]
                    history[this]=dp[this]+history[pair[0]]+history[pair[1]]
                else:
                    orig = history[this]
                    dp_temp=dp[pair[0]]+dp[pair[1]]
                    history[this]=min(orig, dp_temp+history[pair[0]]+history[pair[1]])
                    if history[this]!=orig:
                        dp[this]=dp_temp
    return history['1'*N]

def separate(a,b):
    front = []
    back = []
    for i in range(b-a):
        front.append((a,a+i))
        back.append((a+i+1, b))
    return front, back

def solve(N, parts):
    added = [0]
    for i in range(len(parts)):
        added.append(parts[i]+added[i])

    dp = []
    for i in range(N):
        temp = [0]*N
        dp.append(temp)

    for window in range(2, N+1):
        for i in range(N-window+1):
            front, back = separate(i, i+window-1)
            dp[i][i+window-1]=float('inf')
            for j in range(len(front)):
                f=front[j]
                b=back[j]
                dp[i][i+window-1]=min(dp[i][i+window-1], added[i+window]-added[i]+dp[f[0]][f[1]]+dp[b[0]][b[1]])
    return dp[0][N-1]

sys.stdin = open("sample_input_3367.txt", "r")
T = int(sys.stdin.readline())
# T = int(input())
# 여러개의 테스트 케이스가 주어지므로, 각각을 처리합니다.
# print(T)
for test_case in range(1, T + 1):
    N = int(sys.stdin.readline())
    parts = list(map(int, sys.stdin.readline().split()))
    # if test_case in [1]:
    #     continue
    ans = solve(N, parts)
    print('#{} {}'.format(test_case, ans))
    # break
```

```python
def solve2(K, files):
    # 누적합 계산
    prefix_sum = [0] * (K + 1)
    for i in range(1, K + 1):
        prefix_sum[i] = prefix_sum[i-1] + files[i-1]

    # dp[i][j]: i부터 j까지의 최소 합병 비용
    dp = [[0] * (K + 1) for _ in range(K + 1)]

    for length in range(2, K + 1):  # 부분 구간 길이
        for i in range(1, K - length + 2):  # 시작점
            j = i + length - 1
            dp[i][j] = float('inf')
            total = prefix_sum[j] - prefix_sum[i-1]
            for k in range(i, j):
                cost = dp[i][k] + dp[k+1][j] + total
                dp[i][j] = min(dp[i][j], cost)
    return dp[1][K]
```
