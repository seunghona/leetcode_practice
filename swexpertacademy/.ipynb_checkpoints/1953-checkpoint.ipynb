```python
import sys
from collections import deque

direcs={0:(-1,0), 1:(0,1), 2:(1,0), 3:(0,-1)}

pipe_d={1: [0,1,2,3],
         2: [0,2],
         3: [1,3],
         4: [0,1],
         5: [1,2],
         6: [2,3],
         7: [3,0]}

def solve(N, M, R, C, L, grid):
    if L==1:
        return 1
    start_pipe = grid[R][C]
    possible = set()
    possible.add((R,C))
    q = deque()
    for d in pipe_d[start_pipe]:
        next_pos = (R+direcs[d][0], C+direcs[d][1])
        q.append((next_pos, d, 2))

    while q:
        # print(possible)
        pos, in_d, cur_l = q.popleft()
        if pos[0]<0 or pos[0]>=N:
            # print('a')
            continue
        if pos[1]<0 or pos[1]>=M:
            # print('b')
            continue
        if pos in possible:
            continue
        if cur_l > L:
            # print('c')
            continue
        if grid[pos[0]][pos[1]] == 0:
            continue
        if (in_d+2)%4 not in pipe_d[grid[pos[0]][pos[1]]]:
            # print('d')
            continue
        possible.add(pos)
        for d in pipe_d[grid[pos[0]][pos[1]]]:
            next_pos = (pos[0]+direcs[d][0], pos[1]+direcs[d][1])
            q.append((next_pos, d, cur_l+1))
        
    return len(possible)

sys.stdin = open("sample_input_1953.txt", "r")
T = int(sys.stdin.readline())
# T = int(input())
# 여러개의 테스트 케이스가 주어지므로, 각각을 처리합니다.
# print(T)
for test_case in range(1, T + 1):
    N, M, R, C, L = map(int, sys.stdin.readline().split())
    grid = []
    for i in range(N):
        temp = list(map(int, sys.stdin.readline().split()))
        grid.append(temp)
    ans = solve(N, M, R, C, L, grid)
    print('#{} {}'.format(test_case, ans))
    # break
```
