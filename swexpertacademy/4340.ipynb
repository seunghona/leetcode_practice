```python
import sys

sys.stdin = open("sample_input_4340.txt", "r")

from collections import deque
pipe_type={}
pipe_type[1]={3:[1], 1:[3], 0:[2], 2:[0]}
pipe_type[2]={3:[1], 1:[3], 0:[2], 2:[0]}
pipe_type[3]={0:[1,3], 1:[0,2], 2:[1,3], 3:[2,0]}
pipe_type[4]={0:[1,3], 1:[0,2], 2:[1,3], 3:[2,0]}
pipe_type[5]={0:[1,3], 1:[0,2], 2:[1,3], 3:[2,0]}
pipe_type[6]={0:[1,3], 1:[0,2], 2:[1,3], 3:[2,0]}

changes={}
changes[0]=[0,-1]
changes[1]=[1,0]
changes[2]=[0,1]
changes[3]=[-1,0]

def get_pipe_code(x,y,n):
    code = '0'*(n*n)
    num = y*n+x
    code=code[0:num]+'1'+code[num+1:]
    return code

def add_pipe_code(c1, c2):
    added = int(c1, 2)+int(c2,2)
    return bin(added)[2:]

def in_grid(x, y, n):
    if x<0 or x>=n:
        return False
    if y<0 or y>=n:
        return False
    return True

def is_pipe(x, y, grid):
    if grid[y][x] > 0:
        return True
    return False

def is_end(x, y, n, out_direc):
    if x==n-1 and y==n-1 and out_direc==1:
    # if x == 0 and y == 0 and out_direc == 3:
        return True
    return False

def solve(n, grid):
    least_steps = n*n
    x = 0
    y = 0
    steps = 1
    current_pipe = grid[y][x]

    get_pipe_code_dict = {}
    for x_ in range(n):
        for y_ in range(n):
            get_pipe_code_dict[(x_,y_)]=get_pipe_code(x_,y_,n)
    print(get_pipe_code_dict)
    used_pipes = get_pipe_code_dict[(x,y)]
    print(used_pipes)
    available_outs = pipe_type[current_pipe][3] #3 is incoming direction
    q = deque()
    visited = {}
    for i in range(len(available_outs)):
        print('hello')
        q.append((x, y, 3, available_outs[i], steps, used_pipes))

    while q:
        oldx, oldy, _, out_direc, steps, used = q.popleft()
        if (oldx, oldy, out_direc, used) not in visited:
            visited[(oldx, oldy, out_direc, used)]=steps
        else:
            prev_steps = visited[(oldx, oldy, out_direc, used)]
            if steps > prev_steps:
                continue

        # if is_end(oldx, oldy, n, out_direc):
        #     if steps < least_steps:
        #         least_steps = steps

        x = oldx + changes[out_direc][0]
        y = oldy + changes[out_direc][1]

        if in_grid(x, y, n) and is_pipe(x, y, grid):
            newsteps=steps+1
            in_direc = (out_direc+2)%4
            available_outs = pipe_type[grid[y][x]][in_direc]
            for i in range(len(available_outs)):
                if is_end(x, y, n, available_outs[i]):
                    if newsteps < least_steps:
                        least_steps = newsteps
                else:
                    newcode = get_pipe_code_dict[(x,y)]
                    updateused = add_pipe_code(used, newcode)
                    q.append((x, y, in_direc, available_outs[i], newsteps, updateused))

    return least_steps

# T = int(input())
T = int(sys.stdin.readline())
# 여러개의 테스트 케이스가 주어지므로, 각각을 처리합니다.
for test_case in range(1, T + 1):
    n = int(sys.stdin.readline())
    grid=[]
    for i in range(int(n)):
        grid.append(list(map(int, sys.stdin.readline().split())))
    least_steps = solve(n, grid)
    print('#{} {}'.format(test_case, least_steps))
    break
```
