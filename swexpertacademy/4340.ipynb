```python
import sys

sys.stdin = open("inputs/sample_input_4340.txt", "r")

from collections import deque
pipe_type={}
pipe_type[1]={3:[1], 1:[3], 0:[2], 2:[0]}
pipe_type[2]={3:[1], 1:[3], 0:[2], 2:[0]}
pipe_type[3]={0:[1,3], 1:[0,2], 2:[1,3], 3:[2,0]}
pipe_type[4]={0:[1,3], 1:[0,2], 2:[1,3], 3:[2,0]}
pipe_type[5]={0:[1,3], 1:[0,2], 2:[1,3], 3:[2,0]}
pipe_type[6]={0:[1,3], 1:[0,2], 2:[1,3], 3:[2,0]}

changes={}
changes[0]=[0,-1]
changes[1]=[1,0]
changes[2]=[0,1]
changes[3]=[-1,0]

def get_pipe_code(x,y,n):
    num = y*n+x
    return bin(1<<num)

def add_pipe_code(c1, c2):
    added = int(c1,2) | int(c2,2)
    return bin(added)

def in_grid(x, y, n):
    if x<0 or x>=n:
        return False
    if y<0 or y>=n:
        return False
    return True

def is_pipe(x, y, grid):
    if grid[y][x] > 0:
        return True
    return False

def is_end(x, y, n, out_direc):
    if x==n-1 and y==n-1 and out_direc==1:
    # if x == 0 and y == 0 and out_direc == 3:
        return True
    return False

def is_start(x, y, n, out_direc):
    if x == 0 and y == 0 and out_direc == 3:
        return True
    return False

def solve(n, grid):
    least_steps = n*n
    x = 0
    y = 0
    steps = 1
    current_pipe = grid[y][x]
    used_pipes = set()
    used_pipes.add((x,y))

    available_outs = pipe_type[current_pipe][3] #3 is incoming direction
    q = deque()
    visited = {}
    for i in range(len(available_outs)):
        q.append((x, y, 3, available_outs[i], steps, used_pipes))

    while q:
        oldx, oldy, in_direc, out_direc, steps, used_pipes = q.popleft()
        if (oldx, oldy, in_direc, out_direc) not in visited:
            visited[(oldx, oldy, in_direc, out_direc)]=steps
        else:
            prev_steps = visited[(oldx, oldy, in_direc, out_direc)]
            if steps > prev_steps:
                continue

        # if is_end(oldx, oldy, n, out_direc):
        #     if steps < least_steps:
        #         least_steps = steps

        x = oldx + changes[out_direc][0]
        y = oldy + changes[out_direc][1]
        if (x,y) in used_pipes:
            continue

        new_used_pipes = used_pipes | {(x,y)}

        if in_grid(x, y, n) and is_pipe(x, y, grid):
            newsteps=steps+1
            in_direc = (out_direc+2)%4
            available_outs = pipe_type[grid[y][x]][in_direc]
            for i in range(len(available_outs)):
                if is_end(x, y, n, available_outs[i]):
                    if newsteps < least_steps:
                        least_steps = newsteps
                else:
                    q.append((x, y, in_direc, available_outs[i], newsteps, new_used_pipes))

    return least_steps

def solve_back(n, grid):
    least_steps = n*n
    x = n-1
    y = n-1
    steps = 1
    current_pipe = grid[y][x]
    used_pipes = set()
    used_pipes.add((x,y))

    available_outs = pipe_type[current_pipe][1] #1 is end direction
    q = deque()
    visited = {}
    for i in range(len(available_outs)):
        q.append((x, y, 1, available_outs[i], steps, used_pipes))

    while q:
        oldx, oldy, in_direc, out_direc, steps, used_pipes = q.popleft()
        if (oldx, oldy, in_direc, out_direc) not in visited:
            visited[(oldx, oldy, in_direc, out_direc)]=steps
        else:
            prev_steps = visited[(oldx, oldy, in_direc, out_direc)]
            if steps > prev_steps:
                continue

        # if is_end(oldx, oldy, n, out_direc):
        #     if steps < least_steps:
        #         least_steps = steps

        x = oldx + changes[out_direc][0]
        y = oldy + changes[out_direc][1]
        if (x,y) in used_pipes:
            continue

        new_used_pipes = used_pipes | {(x,y)}

        if in_grid(x, y, n) and is_pipe(x, y, grid):
            newsteps=steps+1
            in_direc = (out_direc+2)%4
            available_outs = pipe_type[grid[y][x]][in_direc]
            for i in range(len(available_outs)):
                if is_start(x, y, n, available_outs[i]):
                    if newsteps < least_steps:
                        least_steps = newsteps
                else:
                    q.append((x, y, in_direc, available_outs[i], newsteps, new_used_pipes))

    return least_steps

# T = int(input())
T = int(sys.stdin.readline())
# 여러개의 테스트 케이스가 주어지므로, 각각을 처리합니다.
for test_case in range(1, T + 1):
    n = int(sys.stdin.readline())
    grid=[]
    for i in range(int(n)):
        grid.append(list(map(int, sys.stdin.readline().split())))
    least_steps = solve(n, grid)
    least_steps_back = solve_back(n, grid)
    print(least_steps, least_steps_back)
    print('#{} {}'.format(test_case, min(least_steps, least_steps_back)))
```
