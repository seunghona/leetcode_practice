```python
import sys

def check(D, W, K, film):
    # print(film)
    check_c = 0
    for w in range(W):
        prev = check_c
        for d in range(D-K+1):
            column=[]
            for k in range(K):
                column.append(film[d+k][w])
            
            checksum = sum(column)
            # print(column, check_c)
            # print(checksum)
            if checksum==K or checksum==0:
                check_c+=1
                break
        if prev==check_c:
            return False
    # if check_c!=W:
    #     return False
    # print('True')
    return True

def solve(D, W, K, film):
    ans = float('inf')
    def dfs(film, last_depth, injections):
        # print(film)
        nonlocal ans
        # check if satisfies
        #     if satisfy, update ans=min(ans, injections)
        # if not
        #     for last_depth~depth
        #         dfs(film_inject, last_depth+j, injections+1)
        if injections>K:
            return
            
        satisfies = check(D, W, K, film)
        if satisfies:
            ans=min(ans, injections)
            # print(ans)
            return
        
        if last_depth==D-1:
            return
        
        for d in range(last_depth+1, D):
            zeros=[0]*W
            dfs(film[0:d]+[zeros]+film[d+1:], d, injections+1)
            ones=[1]*W
            dfs(film[0:d]+[ones]+film[d+1:], d, injections+1)
    dfs(film, -1, 0)
        
    return ans
        

sys.stdin = open("sample_input_2112.txt", "r")
T = int(sys.stdin.readline())
# T = int(input())
# 여러개의 테스트 케이스가 주어지므로, 각각을 처리합니다.
# print(T)
for test_case in range(1, T + 1):
    D, W, K = map(int, sys.stdin.readline().split())
    film = []
    for i in range(D):
        temp = list(map(int, sys.stdin.readline().split()))
        film.append(temp)
    if test_case in []:
        continue
    # print(D, W, K)
    ans = solve(D, W, K, film)
    print('#{} {}'.format(test_case, ans))
    # break
```
