```python
import sys

direcs = {1:(-1,0), 2: (1,0), 3:(0,-1), 4:(0,1)}

class Cell:
    def __init__(self, pos, num, direc):
        self.pos = pos
        self.num = num
        self.direc = direc
        
    def printc(self):
        print(self.pos)
        print(self.num)
        print(self.direc)
        
    def move(self, N):
        self.pos=(self.pos[0]+direcs[self.direc][0], self.pos[1]+direcs[self.direc][1])
        if self.pos[0]==N-1 or self.pos[0]==0 or self.pos[1]==N-1 or self.pos[1]==0:
            if self.num%2==0:
                self.num=int((self.num)/2)
            else:
                self.num=int((self.num-1)/2)
            if self.direc==1:
                self.direc=2
            elif self.direc==2:
                self.direc=1
            elif self.direc==3:
                self.direc=4
            elif self.direc==4:
                self.direc=3

def solve(N, M, K, cells):
    for i in range(M):
        # print('')
        # move cells
        #     if they are at boundaries, divide and change direc
        # check if they are at same position
        #     if same position, add into one cell
        pos_sum = {}
        for cell in cells:
            cell.move(N)
            if cell.num==0:
                continue
            pos = cell.pos
            if pos in pos_sum:
                pos_sum[pos].append(cell)
            else:
                pos_sum[pos]=[cell]

        for pos in pos_sum.keys():
            if len(pos_sum[pos])>1:
                idx=0
                comp=pos_sum[pos][0].num
                d=pos_sum[pos][0].direc
                num_sum = pos_sum[pos][0].num
                for j in range(1, len(pos_sum[pos])):
                    num_sum+=pos_sum[pos][j].num
                    if pos_sum[pos][j].num>comp:
                        idx=j
                        comp=pos_sum[pos][j].num
                        d=pos_sum[pos][j].direc
                newcell = Cell(pos, num_sum, d)
                pos_sum[pos]=[newcell]

        cells = []
        for pos in pos_sum.keys():
            cells.append(pos_sum[pos][0])
            # pos_sum[pos][0].printc()
    tot=0
    for cell in cells:
        tot+=cell.num
        
    return tot

sys.stdin = open("sample_input_2382.txt", "r")
T = int(sys.stdin.readline())
# T = int(input())
# 여러개의 테스트 케이스가 주어지므로, 각각을 처리합니다.
# print(T)
for test_case in range(1, T + 1):
    N, M, K = map(int, sys.stdin.readline().split())
    cells = []
    for i in range(K):
        temp = list(map(int, sys.stdin.readline().split()))
        cells.append(Cell((temp[0], temp[1]), temp[2], temp[3]))
    # if test_case in [1]:
    #     continue
    ans = solve(N, M, K, cells)
    print('#{} {}'.format(test_case, ans))
    # break
```
