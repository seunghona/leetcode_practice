```python
import sys

def solve(N, weights):
    cnt = 0
    def dfs(left, right, remaining, diff):
        nonlocal cnt
        if len(remaining)==0:
            cnt+=1
            return
        for i in range(len(remaining)):
            weight_c = remaining[i]
            remaining_new = remaining[0:i]+remaining[i+1:]

            dfs(left+str(weight_c), right, remaining_new, diff+weight_c)
            if weight_c <= diff:
                dfs(left, right+str(weight_c), remaining_new, diff-weight_c)
    dfs('', '', weights, 0)
    
    return cnt

def solve2(N, weights):
    weights.sort()
    dp={}
    def dfs(left, right, remaining, diff):
        nonlocal dp
        str_remaining = "".join(map(str, remaining))
        if len(remaining)==0:
            dp[(str_remaining, diff)]=1
            return 1
            
        if (str_remaining, diff) in dp.keys():
            return dp[(str_remaining, diff)]
            
        total = 0
        for i in range(len(remaining)):
            weight_c = remaining[i]
            remaining_new = remaining[0:i]+remaining[i+1:]

            total+=dfs(left+str(weight_c), right, remaining_new, diff+weight_c)
            if weight_c <= diff:
                total+=dfs(left, right+str(weight_c), remaining_new, diff-weight_c)
        dp[(str_remaining, diff)]=total
        return total
    cnt = dfs('', '', weights, 0)
    
    return cnt

from itertools import permutations

def count_valid_ways(weights):
    N = len(weights)
    cnt = 0
    
    def dfs(idx, left_sum, right_sum, perm):
        nonlocal cnt
        if idx == N:
            cnt += 1
            return
        
        weight = perm[idx]
        
        # 왼쪽에 올리는 경우는 항상 가능
        dfs(idx + 1, left_sum + weight, right_sum, perm)
        
        # 오른쪽에 올리는 경우는 조건을 만족해야 가능
        if right_sum + weight <= left_sum:
            dfs(idx + 1, left_sum, right_sum + weight, perm)
    
    for perm in permutations(weights):  # N! 순열 생성
        dfs(0, 0, 0, perm)
    
    return cnt

def solve_chat(weights):
    N = len(weights)
    max_weight = sum(weights)
    max_left = 10000  # practical cap to left weight

    dp = [[[-1 for _ in range(1 << N)] for _ in range(max_left + 1)] for _ in range(N + 1)]

    def dfs(depth, left, right, visited):
        if right > left:
            return 0
        if depth == N:
            return 1
        if dp[depth][left][visited] != -1:
            return dp[depth][left][visited]

        total = 0
        for i in range(N):
            if not (visited & (1 << i)):
                w = weights[i]
                next_visited = visited | (1 << i)

                # 왼쪽에 올리기
                total += dfs(depth + 1, left + w, right, next_visited)

                # 오른쪽에 올리기 (조건 만족 시)
                if right + w <= left:
                    total += dfs(depth + 1, left, right + w, next_visited)

        dp[depth][left][visited] = total
        return total

    return dfs(0, 0, 0, 0)

sys.stdin = open("sample_input_3234.txt", "r")
T = int(sys.stdin.readline())
for test_case in range(1, T + 1):
    N = int(sys.stdin.readline())
    weights = list(map(int, sys.stdin.readline().split()))
    ans = solve2(N, weights)
    # ans = count_valid_ways(weights)
    # ans = solve_chat(weights)
    print('#{} {}'.format(test_case, ans))
    # break
```
