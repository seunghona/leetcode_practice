```python
import sys

class Cell:
    def __init__(self, pos, num, timeleft, state):
        self.pos = pos
        self.num = num
        self.timeleft = timeleft
        self.state = state
    def printcell(self):
        print(self.pos)
        print(self.num)
        print(self.timeleft)
        print(self.state)

def solve(N, M, K, cells):
    visited = set()
    for cell in cells:
        visited.add(cell.pos)
    for i in range(1, K+1):
        # go through cells
        # state changes to check
        # at 0 (dormant), decrease timeleft. if timeleft is 0, change current state to 1.
        # at 1 (active), decrease timeleft. infect surrounding cells. if timeleft is 0, change state to dead.
        # at -1 (dead), just dead.
        newcells = {}
        for cell in cells:
            # cell.printcell()
            state = cell.state
            if state == -1:
                continue
            x, y = cell.pos
            cell.timeleft -= 1
            if state == 1:
                newposs = [(x,y+1), (x,y-1), (x+1,y), (x-1,y)]
                for newpos in newposs:
                    if newpos not in visited:
                        c = Cell(newpos, cell.num, cell.num, 0)
                        if newpos not in newcells:
                            newcells[newpos]=c
                        else:
                            prev = newcells[newpos]
                            if c.num > prev.num:
                                newcells[newpos]=c
                if cell.timeleft==0:
                    cell.state=-1
            elif state == 0:
                if cell.timeleft==0:
                    cell.state=1
                    cell.timeleft=cell.num
        for newcell in newcells.values():
            cells.append(newcell)
            visited.add(newcell.pos)
    cnt = 0
    for cell in cells:
        # cell.printcell()
        if cell.state>=0:
            cnt+=1
    return cnt
        
'''
      아래의 구문은 input.txt 를 read only 형식으로 연 후,
      앞으로 표준 입력(키보드) 대신 input.txt 파일로부터 읽어오겠다는 의미의 코드입니다.
      여러분이 작성한 코드를 테스트 할 때, 편의를 위해서 input.txt에 입력을 저장한 후,
      아래 구문을 이용하면 이후 입력을 수행할 때 표준 입력 대신 파일로부터 입력을 받아올 수 있습니다.

      따라서 테스트를 수행할 때에는 아래 주석을 지우고 이 구문을 사용하셔도 좋습니다.
      아래 구문을 사용하기 위해서는 import sys가 필요합니다.

      단, 채점을 위해 코드를 제출하실 때에는 반드시 아래 구문을 지우거나 주석 처리 하셔야 합니다.
'''
sys.stdin = open("sample_input_5653.txt", "r")
T = int(sys.stdin.readline())
# T = int(input())
# 여러개의 테스트 케이스가 주어지므로, 각각을 처리합니다.
# print(T)
for test_case in range(1, T + 1):
    N, M, K = map(int, sys.stdin.readline().split())
    cells = []
    for i in range(N):
        temp = list(map(int, sys.stdin.readline().split()))
        for j in range(len(temp)):
            if temp[j]>0:
                cell = Cell((i,j), temp[j], temp[j], 0)
                cells.append(cell)
    ans = solve(N, M, K, cells)
    print('#{} {}'.format(test_case, ans))
    # break
```
