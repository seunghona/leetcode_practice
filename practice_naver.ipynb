```python
def solution(indices, K):
    # Implement your solution here
    dataset_len = len(indices) // K
    remaining = len(indices) % K
    test_lens = [dataset_len + 1 if i < remaining else dataset_len for i in range(K)]
    # print(test_lens)
    test_lens_accuml = []
    for i in range(len(test_lens)+1):
        test_lens_accuml.append(sum(test_lens[:i]))
    # print(test_lens_accuml)
    
    answer = []
    for i in range(K):
        if i<K-1:
            test = indices[test_lens_accuml[i]: test_lens_accuml[i+1]]
            trainfront = indices[0: test_lens_accuml[i]]
            trainback = indices[test_lens_accuml[i+1]:]
        else:
            test = indices[test_lens_accuml[i]:]
            trainfront = indices[0: test_lens_accuml[i]]
            trainback = []
        trainfront.extend(trainback)
        answer.append(trainfront)
        answer.append(test)
    return answer
```
