```python
# https://leetcode.com/problems/dota2-senate/?envType=study-plan-v2&envId=leetcode-75
def predictPartyVictory(senate: str) -> str:
    p1 = senate[0]
    if p1=='R':
        p2='D'
        p1f='Radiant'
        p2f='Dire'
    else:
        p2='R'
        p1f='Dire'
        p2f='Radiant'
    p1count = 1
    p2skip = 1
    
    p2count = 0
    p1skip = 0

    front = p1f
    for i in range(1, len(senate)):
        print(p1count, p1skip, p2count, p2skip)
        """
        p1, p1skip pos
        p1, p1skip 0
        p2, p2skip 0
        p2, p2skip pos
        """
        if senate[i]==p1 and p1skip>0:
            print('1')
            p1skip-=1
        elif senate[i]==p1 and p1skip==0:
            print('2')
            p1count+=1
            p2skip+=1
            if p2count==p2skip and p2count>1:
                front=p1f
        elif senate[i]==p2 and p2skip==0:
            print('3')
            p2count+=1
            p1skip+=1
            if p1count==p1skip and p2count>1:
                front=p2f
        else:
            print('4')
            p2skip-=1
        
    if p1count-p1skip>p2count-p2skip:
        return p1f
    elif p1count-p1skip==p2count-p2skip:
        return front
    return p2f
```

```python
senate = "RRDRDDRDRRDDDDDRDRDR"
print(len(senate))
count=0
for i in senate:
    if i=='R':
        count+=1
print(count)
predictPartyVictory(senate)
```
