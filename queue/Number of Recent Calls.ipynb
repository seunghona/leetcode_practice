```python
# https://leetcode.com/problems/number-of-recent-calls/?envType=study-plan-v2&envId=leetcode-75
class RecentCounter:

    def __init__(self):
        self.count=0
        self.time=[]
        self.low_idx=0

    def ping(self, t: int) -> int:
        self.time.append(t)
        if len(self.time)==1:
            self.low_idx==0

        if self.time[self.low_idx] >= t-3000:
            return len(self.time)-self.low_idx
        else:
            while self.time[self.low_idx] < t-3000:
                self.low_idx+=1
                if self.low_idx >= len(self.time):
                    return 1
            return len(self.time)-self.low_idx
            


# Your RecentCounter object will be instantiated and called as such:
# obj = RecentCounter()
# param_1 = obj.ping(t)
```
